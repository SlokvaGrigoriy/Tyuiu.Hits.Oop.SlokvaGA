**Описание задачи:**

<p style="text-align: justify;">Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.</p>

<b>Требования к базовому классу Supplier</b>:
<ul type="I">
    <li><b>Атрибуты:</b> ID поставщика (SupplierId), Название компании (CompanyName),<br>
    Тип продукции (ProductType).
    </li>
    <li><b>Методы:</b>
        <ul type="I">
            <li>GetCompanyInfo(): метод для получения информации о компании.</li>
            <li>ProvideQuote(): метод для предоставления котировки на товары или<br>
            услуги.</li>
            <li>SubmitOrder(): метод для отправки заказа поставщику.</li>
        </ul>
    </li>
</ul>
<b>Требования к производным классам:</b>
    <ol>
    <li><p style="text-align: justify;">Производитель (Manufacturer): Должен содержать дополнительные атрибуты,
    такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть
    переопределен для включения информации о годе основания компании в
    котировку.<p></li>
    <li><p style="text-align: justify;">Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как
    Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть
    переопределен для добавления информации о расположении магазина при
    отправке заказа.</p></li>
    <li><p style="text-align: justify;">Импортер (Importer) (если требуется третий класс): Должен содержать
    дополнительные атрибуты, такие как Страна происхождения товара
    (OriginCountry). Метод GetCompanyInfo() должен быть переопределен для
    отображения страны происхождения товара вместе с остальной
    информацией о компании.</p></li>
    </ol>

#### Дополнительное задание
Добавьте к существующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и перегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public class Supplier
{
    private string _supplierId;
    private string _companyName;
    private string _productType;

    public string SupplierId 
    { 
        get { return _supplierId; } 
        set { _supplierId = string.IsNullOrWhiteSpace(value) ? "Не указан" : value; } 
    }
    
    public string CompanyName 
    { 
        get { return _companyName; } 
        set { _companyName = string.IsNullOrWhiteSpace(value) ? "Без названия" : value; } 
    }
    
    public string ProductType 
    { 
        get { return _productType; } 
        set { _productType = string.IsNullOrWhiteSpace(value) ? "Не определен" : value; } 
    }

    public double Rating { get; set; } = 5.0;

    public Supplier(string supplierId, string companyName, string productType)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID: {SupplierId}, Компания: {CompanyName}, Тип: {ProductType}, Рейтинг: {Rating}⭐");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine($"Стандартная котировка от {CompanyName}.");
    }

    public virtual void ProvideQuote(double discount)
    {
        Console.WriteLine($"Котировка от {CompanyName} со скидкой {discount}%.");
    }

    public virtual void ProvideQuote(string urgentMode)
    {
        Console.WriteLine($"Срочная котировка от {CompanyName} в режиме: {urgentMode}.");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}.");
    }
}

public interface IDeliverable
{
    void ShipGoods();
}

public class Manufacturer : Supplier
{
    private int _foundedYear;

    public int FoundedYear 
    { 
        get { return _foundedYear; } 
        set { _foundedYear = value > 0 ? value : 0; } 
    }

    public string FactoryAddress { get; set; } = "Промзона";
    public int EmployeesCount { get; set; } = 150;

    public Manufacturer(string supplierId, string companyName, string productType, int foundedYear) 
        : base(supplierId, companyName, productType)
    {
        FoundedYear = foundedYear;
    }

    public override void ProvideQuote()
    {
        string yearText = FoundedYear == 0 ? "Год не определен" : FoundedYear.ToString();
        Console.WriteLine($"Котировка от завода {CompanyName} (Год основания: {yearText}).");
    }

    public void LaunchProduction()
    {
        Console.WriteLine($"Завод {CompanyName} запустил конвейер для выпуска товара: {ProductType}.");
    }
}

public class Retailer : Supplier, IDeliverable
{
    private string _storeLocation;

    public string StoreLocation 
    { 
        get { return _storeLocation; } 
        set { _storeLocation = string.IsNullOrWhiteSpace(value) ? "Онлайн" : value; } 
    }

    public double RetailMarkup { get; set; } = 15.0;
    public string ContactPhone { get; set; } = "+7-999-000-00-00";

    public Retailer(string supplierId, string companyName, string productType, string storeLocation) 
        : base(supplierId, companyName, productType)
    {
        StoreLocation = storeLocation;
    }

    public void ShipGoods()
    {
        Console.WriteLine($"Ритейлер {CompanyName} отправил курьера из магазина по адресу: {StoreLocation}.");
    }

    public override void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен ритейлеру {CompanyName}. Адрес магазина: {StoreLocation}.");
    }

    public void OrderFromManufacturer(Manufacturer manufacturer)
    {
        Console.WriteLine($"\n--- Взаимодействие: {CompanyName} отправляет запрос на завод {manufacturer.CompanyName} ---");
        manufacturer.ProvideQuote();
        Console.WriteLine($"{CompanyName} подтверждает сделку по типу товара '{manufacturer.ProductType}'.");
    }
}

public class Importer : Manufacturer
{
    private string _originCountry;

    public string OriginCountry 
    { 
        get { return _originCountry; } 
        set { _originCountry = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; } 
    }

    public string ImportLicenseNumber { get; set; } = "LIC-777";
    public int DeliveryDays { get; set; } = 14;

    public Importer(string supplierId, string companyName, string productType, int foundedYear, string originCountry) 
        : base(supplierId, companyName, productType, foundedYear)
    {
        OriginCountry = originCountry;
    }

    public override void GetCompanyInfo()
    {
        Console.WriteLine($"ID: {SupplierId}, Компания: {CompanyName}, Тип: {ProductType}, Страна импорта: {OriginCountry}, Год фабрики: {FoundedYear}");
    }

    public void VerifyLicense()
    {
        Console.WriteLine($"Лицензия №{ImportLicenseNumber} на импорт из страны {OriginCountry} успешно проверена.");
    }
}

public class SupplyChainManager<T> where T : Supplier
{
    private List<T> _suppliers = new List<T>();

    public void AddSupplier(T supplier)
    {
        _suppliers.Add(supplier);
        Console.WriteLine($"В цепочку поставок добавлен: {supplier.CompanyName}");
    }

    public void RemoveSupplier(T supplier)
    {
        _suppliers.Remove(supplier);
    }

    public void ProcessQuotes()
    {
        Console.WriteLine("\n--- Обработка котировок менеджером цепочки поставок ---");
        foreach (var supplier in _suppliers)
        {
            supplier.ProvideQuote();
        }
    }
}

Manufacturer prod = new Manufacturer("M01", "ТехноЗавод", "Станки", 1995);
Retailer shop = new Retailer("R02", "МегаМаркет", "Продукты", "ул. Ленина, 10");
Importer imp = new Importer("I03", "ГлобалТрейд", "Фрукты", 2010, "Бразилия");

Console.WriteLine("--- Вызов базовых и переопределенных методов ---");
prod.ProvideQuote();
shop.SubmitOrder();
imp.GetCompanyInfo();
shop.OrderFromManufacturer(prod);
prod.LaunchProduction();
shop.ShipGoods();
imp.VerifyLicense();

Console.WriteLine("\n--- Демонстрация ПЕРЕГРУЗКИ методов (Overloading) ---");
prod.ProvideQuote(10.5);
prod.ProvideQuote("Экспресс");

Console.WriteLine("\n--- Демонстрация работы GENERIC класса ---");
SupplyChainManager<Supplier> manager = new SupplyChainManager<Supplier>();
manager.AddSupplier(prod);
manager.AddSupplier(shop);
manager.AddSupplier(imp);

manager.ProcessQuotes();
```
