**Описание задачи:**

<p style="text-align: justify;">Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.</p>

<b>Требования к базовому классу Supplier</b>:
<ul type="I">
    <li><b>Атрибуты:</b> ID поставщика (SupplierId), Название компании (CompanyName),<br>
    Тип продукции (ProductType).
    </li>
    <li><b>Методы:</b>
        <ul type="I">
            <li>GetCompanyInfo(): метод для получения информации о компании.</li>
            <li>ProvideQuote(): метод для предоставления котировки на товары или<br>
            услуги.</li>
            <li>SubmitOrder(): метод для отправки заказа поставщику.</li>
        </ul>
    </li>
</ul>
<b>Требования к производным классам:</b>
    <ol>
    <li><p style="text-align: justify;">Производитель (Manufacturer): Должен содержать дополнительные атрибуты,
    такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть
    переопределен для включения информации о годе основания компании в
    котировку.<p></li>
    <li><p style="text-align: justify;">Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как
    Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть
    переопределен для добавления информации о расположении магазина при
    отправке заказа.</p></li>
    <li><p style="text-align: justify;">Импортер (Importer) (если требуется третий класс): Должен содержать
    дополнительные атрибуты, такие как Страна происхождения товара
    (OriginCountry). Метод GetCompanyInfo() должен быть переопределен для
    отображения страны происхождения товара вместе с остальной
    информацией о компании.</p></li>
    </ol>


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
class Supplier
{
    public string SupplierId { get; set; }
    public string CompanyName { get; set; }
    public string ProductType { get; set; }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID: {SupplierId}, Компания: {CompanyName}, Тип: {ProductType}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine($"Стандартная котировка от {CompanyName}.");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}.");
    }
}

class Manufacturer : Supplier
{
    public int FoundedYear { get; set; }

    public override void ProvideQuote()
    {
        Console.WriteLine($"Котировка от завода {CompanyName} (Год основания: {FoundedYear}).");
    }
}

class Retailer : Supplier
{
    public string StoreLocation { get; set; }

    public override void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен ритейлеру {CompanyName}. Адрес магазина: {StoreLocation}.");
    }
}

class Importer : Supplier
{
    public string OriginCountry { get; set; }

    public override void GetCompanyInfo()
    {
        Console.WriteLine($"ID: {SupplierId}, Компания: {CompanyName}, Тип: {ProductType}, Страна импорта: {OriginCountry}");
    }
}

Manufacturer prod = new Manufacturer { SupplierId = "M01", CompanyName = "ТехноЗавод", ProductType = "Станки", FoundedYear = 1995 };
Retailer shop = new Retailer { SupplierId = "R02", CompanyName = "МегаМаркет", ProductType = "Продукты", StoreLocation = "ул. Ленина, 10" };
Importer imp = new Importer { SupplierId = "I03", CompanyName = "ГлобалТрейд", ProductType = "Фрукты", OriginCountry = "Бразилия" };

prod.ProvideQuote();
shop.SubmitOrder();
imp.GetCompanyInfo();
```
